```python
#function that saves fixture data.
def save_fixture_json(fixture_id, fixture_data):
    file_path = RAW_DIR / f"fixture_{fixture_id}.json"

    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(
            fixture_data,
            file,
            ensure_ascii=False,
            indent=2
        )

    return file_path
```

```python
# function that checks whether a 200 response looks complete
def classify_fixture_data(fixture_data):
    if not isinstance(fixture_data, dict):
        return "unexpected_structure" #the response is not a dictionary, so it is structurally unexpected (might have to do some changes to formatting of these)

    if "fixture" not in fixture_data:
        return "unexpected_structure" #the fixture section key is absenrt in the dictionary, so no information like id (our numeric identifier), date, stadium/venue, it is again structurally unexpected (imagine missing dates and venues).

    if fixture_data["fixture"] is None: #fixture key is present, but no information under it, so it is incomplete (will have to check HOW incomplete - it should have date atleast. otherwise drop!)
        return "incomplete"

    if not fixture_data["fixture"].get("id"): #the fixture key exists, but has no id key in its dictionary. We also call that incomplete.
        return "incomplete"

    return "success"
```

```python
def scrape_one_fixture(fixture_id):
    url = f"https://www.the-aiff.com/api/fixtures/cms/{fixture_id}"

    try:
        response = requests.get(url, timeout=30)

    except requests.RequestException as error:
        return {
            "fixture_id": fixture_id,
            "status": "request_error",
            "http_status": "",
            "file_name": "",
            "error": str(error)
        }

    if response.status_code != 200:
        return {
            "fixture_id": fixture_id,
            "status": f"failed_{response.status_code}",
            "http_status": response.status_code,
            "file_name": "",
            "error": response.text[:200]
        }

    try:
        fixture_data = response.json()

    except ValueError as error:
        return {
            "fixture_id": fixture_id,
            "status": "invalid_json",
            "http_status": response.status_code,
            "file_name": "",
            "error": str(error)
        }

    status = classify_fixture_data(fixture_data)

    file_path = save_fixture_json(
        fixture_id,
        fixture_data
    )

    return {
        "fixture_id": fixture_id,
        "status": status,
        "http_status": response.status_code,
        "file_name": file_path.name,
        "error": ""
    }
```

```python
def scrape_one_fixture(fixture_id, max_attempts=3):
    url = f"https://www.the-aiff.com/api/fixtures/cms/{fixture_id}"

    for attempt in range(1, max_attempts + 1):

        try:
            response = requests.get(url, timeout=30)

        except requests.RequestException as error:
            if attempt < max_attempts:
                time.sleep(2)
                continue

            return {
                "fixture_id": fixture_id,
                "status": "request_error",
                "http_status": "",
                "file_name": "",
                "error": str(error)
            }

        if response.status_code == 200:
            try:
                fixture_data = response.json()

            except ValueError as error:
                return {
                    "fixture_id": fixture_id,
                    "status": "invalid_json",
                    "http_status": 200,
                    "file_name": "",
                    "error": str(error)
                }

            status = classify_fixture_data(fixture_data)
            file_path = save_fixture_json(fixture_id, fixture_data)

            return {
                "fixture_id": fixture_id,
                "status": status,
                "http_status": 200,
                "file_name": file_path.name,
                "error": ""
            }

        if attempt < max_attempts:
            time.sleep(2)

    return {
        "fixture_id": fixture_id,
        "status": f"failed_{response.status_code}",
        "http_status": response.status_code,
        "file_name": "",
        "error": response.text[:200]
    }
```

```python
def scrape_one_fixture(fixture_id, max_attempts=3):
    url = f"https://www.the-aiff.com/api/fixtures/cms/{fixture_id}"

    for attempt in range(1, max_attempts + 1):

        try:
            response = requests.get(url, timeout=30)

        except requests.RequestException as error:
            if attempt < max_attempts:
                time.sleep(1)
                continue

            return {
                "fixture_id": fixture_id,
                "status": "request_error",
                "http_status": "",
                "file_name": "",
                "error": str(error),
                "attempts": attempt
            }

        if response.status_code == 200:
            try:
                fixture_data = response.json()

            except ValueError as error:
                return {
                    "fixture_id": fixture_id,
                    "status": "invalid_json",
                    "http_status": 200,
                    "file_name": "",
                    "error": str(error),
                    "attempts": attempt
                }

            status = classify_fixture_data(fixture_data)
            file_path = save_fixture_json(fixture_id, fixture_data)

            return {
                "fixture_id": fixture_id,
                "status": status,
                "http_status": 200,
                "file_name": file_path.name,
                "error": "",
                "attempts": attempt
            }

        if attempt < max_attempts:
            time.sleep(1)

    return {
        "fixture_id": fixture_id,
        "status": f"failed_{response.status_code}",
        "http_status": response.status_code,
        "file_name": "",
        "error": " ".join(response.text[:200].split()),
        "attempts": max_attempts
    }
```

```python
def load_completed_ids():
    completed_ids = set()

    if not LOG_PATH.exists():
        return completed_ids

    with open(LOG_PATH, "r", encoding="utf-8") as log_file:
        reader = csv.DictReader(log_file)

        for row in reader:
            if row["status"] in {
                "success",
                "incomplete",
                "unexpected_structure"
            }:
                completed_ids.add(int(row["fixture_id"]))

    return completed_ids
```
